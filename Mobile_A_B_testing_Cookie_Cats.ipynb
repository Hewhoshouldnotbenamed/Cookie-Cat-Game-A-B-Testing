```python
import pandas as pd
import matplotlib as pt
```

```python
cookie_data=pd.read_csv('/content/cookie_cats.csv')
```

```python
# Cookie data info
print(cookie_data.info())
```

```python
# Comparison of  players retention in terms of stats when for stage30 and stage40 as first.
filtered_data=cookie_data.groupby([cookie_data['version']])
# Stats after different days of retention 1 & 7.
import pandas as pd

# Summary statistics for Day 1 retention by version
# (Using .mean() gives the retention rate, .sum() gives total retained players)
retention_1_stats = cookie_data.groupby('version')['retention_1'].agg(
    total_players='count',
    retained_players='sum',
    retention_rate='mean'
)

print("Day 1 Retention Stats:")
print(retention_1_stats)

# If your dataset also includes Day 7 retention
if 'retention_7' in cookie_data.columns:
    retention_7_stats = cookie_data.groupby('version')['retention_7'].agg(
        total_players='count',
        retained_players='sum',
        retention_rate='mean'
    )
    print("\nDay 7 Retention Stats:")
    print(retention_7_stats)

day_1_retention_rate_diff=retention_1_stats['retention_rate'].diff()
day_7_retention_rate_diff=retention_7_stats['retention_rate'].diff()
print("Day1 retention_rate_difference:",day_1_retention_rate_diff*100)
print('Day2 retention_rate_difference:',day_7_retention_rate_diff*100)
```

```python
#Average number of rounds played by users
print(cookie_data['sum_gamerounds'].median())
```

Final Decision: Moving the gate form level 30 to level 40 has brought the user retention rate down to 0.6% on first day and 0.7% on day 7. The average number of rounds played by users is 16.
